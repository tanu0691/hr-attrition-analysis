```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')

# First look
print("Shape:", df.shape)
print("\nAttrition value counts:")
print(df['Attrition'].value_counts())
print("\nAttrition %:")
print(round(df['Attrition'].value_counts(normalize=True) * 100, 2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Convert Attrition to 1/0
df['Attrition'] = df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

# Check any missing values
print("Missing values:", df.isnull().sum().sum())

# Basic stats
print("\nAverage Age:", round(df['Age'].mean(), 1))
print("Average Monthly Income:", round(df['MonthlyIncome'].mean(), 1))
print("Average Years at Company:", round(df['YearsAtCompany'].mean(), 1))
```

```python
# Chart 1 - Attrition Distribution
plt.figure(figsize=(6,4))
sns.countplot(x='Attrition', data=df, palette='Set2')
plt.title('Employee Attrition Distribution')
plt.xticks([0,1], ['Stayed', 'Left'])
plt.ylabel('Number of Employees')
plt.show()

# Chart 2 - Age vs Attrition
plt.figure(figsize=(8,4))
sns.histplot(data=df, x='Age', hue='Attrition', bins=30, palette='Set2')
plt.title('Age vs Attrition')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.show()
```

```python
# Chart 3 - Department vs Attrition
plt.figure(figsize=(8,4))
sns.countplot(x='Department', hue='Attrition', data=df, palette='Set2')
plt.title('Department vs Attrition')
plt.legend(['Stayed', 'Left'])
plt.ylabel('Number of Employees')
plt.show()
```

```python
# Chart 4 - Job Satisfaction vs Attrition
plt.figure(figsize=(7,4))
sns.countplot(x='JobSatisfaction', hue='Attrition', data=df, palette='Set2')
plt.title('Job Satisfaction vs Attrition')
plt.xlabel('Job Satisfaction (1=Low, 4=High)')
plt.legend(['Stayed', 'Left'])
plt.ylabel('Number of Employees')
plt.show()
```

```python
# Chart 5 - Monthly Income vs Attrition
plt.figure(figsize=(7,4))
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df, palette='Set2')
plt.title('Monthly Income vs Attrition')
plt.xticks([0,1], ['Stayed', 'Left'])
plt.ylabel('Monthly Income ($)')
plt.show()
```

```python
# Chart 6 - Work Life Balance vs Attrition
plt.figure(figsize=(7,4))
sns.countplot(x='WorkLifeBalance', hue='Attrition', data=df, palette='Set2')
plt.title('Work Life Balance vs Attrition')
plt.xlabel('Work Life Balance (1=Bad, 4=Best)')
plt.legend(['Stayed', 'Left'])
plt.ylabel('Number of Employees')
plt.show()
```
